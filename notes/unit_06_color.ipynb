`CET-R3` 只是冰山一角，pyqtgraph 内置了完整的 CET 系列色彩映射。以下是按类别整理的：

### 内置 CET 系列（来自 ColorCET 项目，感知均匀）

| 类别 | 名称 | 用途 |
|------|------|------|
| **Linear (连续)** | `CET-L1` ~ `CET-L20` | 最常用的线性渐变，热力图首选 |
| **Rainbow (彩虹)** | `CET-R1` ~ `CET-R4` | 彩虹色映射（`R3` 就是你用的） |
| **Diverging (发散)** | `CET-D1` ~ `CET-D13` | 有中心零点的双极数据（如正负温差） |
| **Cyclic (循环)** | `CET-CBC1`, `CET-CBC2` | 周期性数据（如角度 0°~360°） |
| **Colorblind (色盲友好)** | `CET-CBL1` ~ `CET-CBL4` | 红绿色盲友好 |
| **Isoluminant** | `CET-I1` ~ `CET-I3` | 等亮度映射 |

### 其他内置

| 名称 | 说明 |
|------|------|
| `viridis` | Matplotlib 默认，感知均匀 |
| `plasma` | 亮色高对比度 |
| `inferno` | 暗色背景适用 |
| `magma` | 类似 inferno |
| `cividis` | 色盲友好 |
| `turbo` | 改进版 jet，Google 开发 |

### 外部来源（需安装对应库）

```python
# 列出所有 Matplotlib colormap（需 pip install matplotlib）
pg.colormap.listMaps('matplotlib')
# → ['viridis', 'plasma', ..., 'hot', 'cool', 'jet', 'RdBu', ...]

# 列出所有 ColorCET colormap（需 pip install colorcet）
pg.colormap.listMaps('colorcet')
```

### 推荐选择

| 场景 | 推荐 |
|------|------|
| 通用热力图 | `CET-L9` 或 `viridis` |
| 暗色背景 | `inferno` 或 `CET-L4` |
| 有正负的数据 | `CET-D1` 或 `RdBu` |
| 帮你用 `CET-R3` | 彩虹色，好看但非感知均匀，科学绘图慎用 |
